```python
df['Date'] = pd.to_datetime(df['Date'])
df.info()
```

```python
# Convert duration columns (HH:MM:SS) to numeric total seconds
duration_cols = ['Answer Speed (AVG)', 'Talk Duration (AVG)', 'Waiting Time (AVG)']

for col in duration_cols:
    df[col] = pd.to_timedelta(df[col]).dt.total_seconds()

# Display the converted columns to verify the result
display(df)
```
